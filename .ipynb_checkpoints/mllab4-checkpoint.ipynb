```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
import kagglehub
path = kagglehub.dataset_download("saurabh00007/iriscsv")
print("Path to dataset files:", path)

data = pd.read_csv("/home/guru/.cache/kagglehub/datasets/saurabh00007/iriscsv/versions/1/Iris.csv")
iris_df = data.drop(columns='Id')
print("Iris Dataset Sample:")
print(iris_df.head())
```

```python
x = iris_df.drop(columns=['Species'])
y = iris_df['Species']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
clf = DecisionTreeClassifier(criterion='entropy',random_state=42)
clf.fit(x_train,y_train)
y_pred = clf.predict(x_test)
accuracy =  accuracy_score(y_test,y_pred)
print("Decision Tree (ID3) Accuracy:",accuracy)
```

```python
new_sample1= pd.DataFrame([[6.6,3.0,4.4,1.4]],columns=['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']) 
prediction1 = clf.predict(new_sample1) 
new_sample2= pd.DataFrame([[5.7,2.5,5.0,2.0]],columns=['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']) 
prediction2 = clf.predict(new_sample2) 
print("Predicted class for new Sample:",prediction1[0]) 
print("Predicted class for new Sample:",prediction2[0])
```
