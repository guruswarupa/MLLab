```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
import kagglehub
path = kagglehub.dataset_download("saurabh00007/iriscsv")
print("Path to dataset files:", path)

data = pd.read_csv("/home/guru/.cache/kagglehub/datasets/saurabh00007/iriscsv/versions/1/Iris.csv")
data = data.drop(columns='Id')

print(data.head())
```

```python
x = data.iloc[:,0:4].values
y = data.iloc[:,4].values
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(x_scaled)
```

```python
x_pca
```

```python
print("Explained variance Ratio:",pca.explained_variance_ratio_)
```

```python
plt.figure(figsize=(8,6))
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_encoded = le.fit_transform(y)
for i in np.unique(y_encoded):
    plt.scatter(x_pca[y_encoded==i,0],
                x_pca[y_encoded==i,1],
                label = le.inverse_transform([i])[0])
plt.xlabel("principal component 1")
plt.ylabel("principal component 2")
plt.title("PCA on Iris dataset(2 components)")
plt.legend()
plt.grid()
plt.show()
print(pca.components_)
    
```
